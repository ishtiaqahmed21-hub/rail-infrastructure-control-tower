# Ledger and source validation

A cumulative weekly ledger must have unique keys, monotonic physical completion, bounded quantities and consistent units. Missing ORR moving averages are retained.

```python
import pandas as pd
d=pd.read_csv('data/processed/ledger.csv')
checks={'duplicate_keys':int(d.duplicated(['wbs','week']).sum()),'negative_cost':int((d.actual_cost<0).sum()),'scope_exceeded':int((d.actual_quantity>d.scope_quantity).sum()),'backward_progress':int((d.groupby('wbs').actual_quantity.diff()<0).sum())}
assert not any(checks.values())
checks
```
